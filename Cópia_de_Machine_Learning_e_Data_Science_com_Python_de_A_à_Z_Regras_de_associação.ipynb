```python
import pandas as pd #leitura arquivo csv
from apyori import apriori #biblioteca instalada - apriori é o algoritmo
```

## Base de dados mercado 1

```python
base_mercado1 = pd.read_csv('mercado.csv', header = None) #header = None - para não considerar o título das colunas
#em formato de dataframe
base_mercado1
#NaN - valor nulo
```

```python
base_mercado1.shape
```

```python
transacoes = [] #em formato de lista
for i in range(len(base_mercado1)):
  #print(i) #-> mostrando os indices
  #print(base_mercado1.values[i, 0]) #_> vendo linha e coluna específica
  transacoes.append([str(base_mercado1.values[i, j]) for j in range(base_mercado1.shape[1])])
  #j - para acessar todas as colunas
  #base_mercado1.shape[1] -> 4 na segunda posição do shape (colunas)
  #[] e str para acessar a lista, deixando em string e em formato de lista ([])
```

```python
transacoes
```

```python
type(transacoes)#vendo o tipo
```

```python
#gerando as regras de associação
regras = apriori(transacoes, min_support = 0.3, min_confidence = 0.8, min_lift = 2)
#parâmetros
#min_support = 0.3 -> suporte mínimo
#min_confidence = 0.8 -> confiança mínima
#min_lift = 2 -> lift mínimo
resultados = list(regras)#transformando a variável regras em lista para melhor visualização
```

```python
#Posição 0 -> itemset
#Posição 1 -> suporte
#Posição 2 -> regras

A = [] #SE
B = [] #ENTAO
suporte = []
confianca = []
lift = []

for resultado in resultados:
  #print(resultado) #mostra os itensset
  s = resultado[1] #valores de suporte
  #print(s)
  result_rules = resultado[2]
  #print(result_rule)
  for result_rule in result_rules: #percorrendo cada uma das regras, todos os itens
    #print(result_rule)
    a = list(result_rule[0]) #se
    b = list(result_rule[1]) #entao
    c = result_rule[2] #confiança
    l = result_rule[3] #lift
    #print(a, ' - ', b, ' - ', c, ' - ', l)
    #adicionando as regras na lista
    A.append(a)
    B.append(b)
    suporte.append(s)
    confianca.append(c)
    lift.append(l)
```

```python
rules_df = pd.DataFrame({'A': A, 'B': B, 'suporte': suporte, 'confianca': confianca, 'lift': lift}) #botando em dataframe
```

```python
rules_df.sort_values(by = 'lift', ascending = False) #ordenando os dados
#.sort_values -> comando de ordenação
#by = 'lift' -> coluna que quero ordenar
#ascending = False -> em ordem decrescente
```

## Base de dados mercado 2

```python
base_mercado2 = pd.read_csv('mercado2.csv', header=None) #base de dados com mais registros
base_mercado2
#NaN -> valor nulo (não comprados)
```

```python
base_mercado2.shape
```

```python
transacoes = [] #transformando em lista
for i in range(base_mercado2.shape[0]):
  transacoes.append([str(base_mercado2.values[i, j]) for j in range(base_mercado2.shape[1])])
```

```python
transacoes
```

```python
# Produtos que são vendidos 4 vezes por dia -> objetivo
```

```python
4 * 7
#7 -> dias da semana
```

```python
28 / 7501
#objetivo/registro total
```

```python
#gerando as regras
regras = apriori(transacoes, min_support = 0.003, min_confidence = 0.2, min_lift = 3)
#suporte bem menro devido a quantidade de registro de produto
resultados = list(regras)
len(resultados)#contando o valor de resultados retornados
```

```python
#Retornando em formato de dataframe
#Posição 0 -> itemset
#Posição 1 -> suporte
#Posição 2 -> regras


A = []
B = []
suporte = []
confianca = []
lift = []

for resultado in resultados:
  s = resultado[1]
  result_rules = resultado[2]
  for result_rule in result_rules:
    a = list(result_rule[0])
    b = list(result_rule[1])
    c = result_rule[2]
    l = result_rule[3]
    A.append(a)
    B.append(b)
    suporte.append(s)
    confianca.append(c)
    lift.append(l)
```

```python
rules_df = pd.DataFrame({'A': A, 'B': B, 'suporte': suporte, 'confianca': confianca, 'lift': lift}) #ordenando os dados
rules_df.sort_values(by = 'confianca', ascending = False)
```

```python
base_mercado1 = pd.read_csv('mercado.csv', header = None) #header = None -> para ignorar o nome das colunas
base_mercado1
```

```python
from pyECLAT import ECLAT
eclat = ECLAT(data = base_mercado1)
#retorno dos objetos mais frequentes
```

```python
eclat.df_bin #gerando um dataframe com os valores binários]
#algoritmo criou o nome das colunas
#possivel forma de representação quando trablha com itens frequentes
```

```python
eclat.uniq_ #retornando o nome das colunas
```

```python
#geração dos índices frequentes
indices, suporte = eclat.fit(min_support=0.3, min_combination=1, max_combination=3)
#não vai combinar em um mesmo itemset menos que 1 produto ou mais que 3 produtos
#ou sej,a suporte para um só produto no mínimo  bno máximo 3
```
